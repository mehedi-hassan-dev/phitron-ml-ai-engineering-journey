## Problem 5: Analyze Passenger Survival
1. Count how many passengers survived and how many did not survive.
2. Calculate the percentage for each group.
3. Create a Seaborn count plot for the Survived column.

```python
survived_and_not_survived_count = df['Survived'].value_counts()
survived_and_not_survived_count
```

```python
caculate_percentage = (df['Survived'].value_counts(normalize= True)* 100).round(3)
caculate_percentage
```

```python
plt.Figure(figsize= (6,4))
sns.countplot(data = df, x='Survived')

plt.title('Passanger survival count (0 = No, 1 = yes)')
plt.xlabel('Survived')
plt.ylabel('Number of Passanger')

plt.show()
```

## Problem 6: Analyze Categorical Features
For each of these columns — Pclass, Sex, and Embarked:

1. Count the number of records in each category.
2. Calculate the percentage of each category.
3. Create a separate Seaborn count plot.

```python
pclass_count = df['Pclass'].value_counts()
sex_count = df['Sex'].value_counts()
embarked_count = df['Embarked'].value_counts()

print(f'Pclass records counts : \n{pclass_count}\n')
print(f'Sex records counts : \n{sex_count}\n')
print(f'Embarked records counts : \n{embarked_count}')
```

```python
pclass_percentage = df['Pclass'].value_counts(normalize=True)*100
sex_percentage = df['Sex'].value_counts(normalize=True)*100
embarked_percentage = df['Embarked'].value_counts(normalize=True)*100

print(f'Pclass records percentage : \n{pclass_percentage.round(2)}\n')
print(f'Sex records percentage : \n{sex_percentage.round(2)}\n')
print(f'Embarked records percentage : \n{embarked_percentage.round(2)}')
```

```python
sns.countplot(data= df, x='Pclass')
plt.title('Pclass Survival Counts')
plt.xlabel('Peaclass')
plt.ylabel('Number of Passanger')

plt.show()
```

```python
sns.countplot(data= df, x='Sex')
plt.title('Sex Survival Counts')
plt.xlabel('Sex')
plt.ylabel('Number of Passanger')

plt.show()
```

```python
sns.countplot(data= df, x='Embarked')
plt.title('Embarked Survival Counts')
plt.xlabel('Embarked')
plt.ylabel('Number of Passanger')

plt.show()
```

## Problem 7: Explore the Fare Distribution
1. Create a histogram of the Fare column using 50 bins.
2. Create a KDE plot for the Fare column.
3. Create a box plot for the Fare column.

```python
plt.Figure(figsize= (6,4))
sns.histplot(data=df, x='Fare', bins= 50)

plt.title('Histogram of passanger Fare')
plt.xlabel('Fare')
plt.ylabel('Frequescy')

plt.show()
```

```python
plt.Figure(figsize= (6,4))
sns.kdeplot(data=df, x='Fare', fill= True)

plt.title('KDE plot of passanger Fare')
plt.xlabel('Fare')
plt.ylabel('Dencity')

plt.show()
```

```python
sns.boxplot(df['Fare'])

plt.title('Box Plot of Passenger Fare')
plt.show()
```
