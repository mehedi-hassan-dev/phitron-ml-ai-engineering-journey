## 1. Countplot এবং Barplot এর মধ্যে পার্থক্য কী? কখন কোনটা use করবে?

### Answer : 
Countplot একটা categorical কলামের প্রতিটা category কতবার এসেছে তা নিজেই গুনে bar দিয়ে দেখায়। এখানে শুধু একটা অক্ষের variable (x বা y) দিতে হয়, আর অন্য অক্ষে আপনাআপনি গণনা (count) বসে যায়।

Barplot প্রতিটা category-র জন্য একটা numeric মানের সারাংশ (ডিফল্টভাবে গড় বা mean) দেখায়। এখানে x-এ category আর y-এ একটা numeric কলাম দিতে হয়। সাধারণত bar-এর মাথায় একটা error bar (confidence interval)-ও থাকে।

বিষয়	Countplot	Barplot
bar-এর উচ্চতা	কতজন/কতবার (frequency)	numeric মানের গড় (বা অন্য সারাংশ)
প্রয়োজনীয় variable	একটা categorical কলাম	একটা categorical + একটা numeric কলাম
গণনা কে করে	Seaborn নিজেই	Seaborn গড় হিসাব করে

কখন কোনটা:

Countplot: যখন জানতে চাই প্রতিটা category-তে কতগুলো data আছে। যেমন: প্রতি department-এ কতজন শিক্ষার্থী।
Barplot: যখন category অনুযায়ী কোনো numeric মান তুলনা করতে চাই। যেমন: প্রতি department-এর গড় score।

## 2. Scatterplot দেখে কীভাবে বুঝবে দুটো variable এর মধ্যে সম্পর্ক আছে কিনা?
### Answer :
Scatterplot-এ প্রতিটা বিন্দু একটা observation, আর বিন্দুগুলো কেমনভাবে সাজানো তা দেখেই সম্পর্ক বোঝা যায়:

দিক (direction):
বিন্দুগুলো বাম-নিচ থেকে ডান-উপরে উঠলে positive সম্পর্ক (একটা বাড়লে অন্যটাও বাড়ে)
ডান-নিচে নামলে negative সম্পর্ক (একটা বাড়লে অন্যটা কমে)
কোনো স্পষ্ট ধারা না থাকলে, এলোমেলো ছড়ানো থাকলে সম্পর্ক নেই বা খুব দুর্বল
শক্তি (strength): বিন্দুগুলো একটা রেখার যত কাছাকাছি জমা থাকে, সম্পর্ক তত শক্তিশালী। যত বেশি ছড়ানো, তত দুর্বল।
ধরন (form): সম্পর্ক সরলরৈখিক (linear) নাকি বাঁকানো (curved) তা দেখা যায়।
Outlier ও cluster: ধারা থেকে অনেক দূরের বিন্দু বা আলাদা গুচ্ছও চোখে পড়ে।

নিশ্চিত হতে correlation coefficient (যেমন df.corr()) বের করা যায়, যা সম্পর্কের শক্তি সংখ্যায় দেয়।

সতর্কতা: সম্পর্ক থাকা মানেই একটা আরেকটার কারণ (causation) নয়।

## 3. YData Profiling কেন manual EDA-র চেয়ে সুবিধাজনক? কী কী information এক জায়গায় পাওয়া যায়?
### Answer :
YData Profiling (আগের নাম pandas-profiling) এমন একটা library যা মাত্র এক লাইন code দিয়ে পুরো dataset-এর একটা স্বয়ংক্রিয় EDA report (সাধারণত HTML) তৈরি করে দেয়।

কেন সুবিধাজনক:

সময় বাঁচে। Manual EDA-তে info(), describe(), isnull().sum(), corr() ও প্রতিটা কলামের chart আলাদা আলাদা করে লিখতে হয়, এখানে সব একসাথে আসে।
কিছু বাদ পড়ার ঝুঁকি কমে। Duplicate, missing value বা সমস্যাযুক্ত কলাম নিজে থেকেই চিহ্নিত হয়।
সব কলামের জন্য একই মানের বিশ্লেষণ হয়, ফলে সামঞ্জস্য থাকে।
Report শেয়ার করা সহজ, কারণ এটা ভিজ্যুয়াল ও গোছানো।

এক জায়গায় যে তথ্যগুলো পাওয়া যায়:

Overview: মোট সারি ও কলাম, missing cell-এর সংখ্যা, duplicate সারি, memory ব্যবহার
Variable-এর ধরন: numeric, categorical, text, datetime ইত্যাদি
প্রতিটা কলামের পরিসংখ্যান: mean, median, std, min, max, quartile, distinct মানের সংখ্যা
Distribution: histogram ও মানের ছড়ানো
Missing value-এর ধরন ও বিন্যাস
Correlation: কলামগুলোর মধ্যে সম্পর্কের ম্যাট্রিক্স
Interaction: দুই variable-এর scatter plot
Alerts (সতর্কবার্তা): high correlation, skewness, অনেক zero, constant কলাম, বেশি unique মান ইত্যাদি
নমুনা data: শুরু ও শেষের কিছু সারি

সীমাবদ্ধতা (যোগ করলে উত্তর আরও পূর্ণাঙ্গ হয়): বড় dataset-এ report বানাতে সময় লাগে, আর এটা গভীর বিশ্লেষণ বা domain-জ্ঞানের বিকল্প নয়। তাই এটাকে শুরুর দ্রুত পর্যালোচনা হিসেবে ব্যবহার করে পরে প্রয়োজনমতো manual EDA করা ভালো।
