# Problem (1) Add New Column and Save File

You have a DataFrame with column Price.   
Create a new column Discounted_Price with (10% discounted price).  
Save the modified DataFrame as discounted_products.csv.


```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv('practice.csv')

df['Discount_price'] = df['Price'] * 0.90

df.to_csv("discounted_products.csv", index = False)

df.head(2)
```

# Problem 02. Check Duplicates and Remove Them
Column: EmployeeID  
 Tasks:  
Count duplicate entries  
Remove duplicates  
Display cleaned DataFrame  


```python
df = pd.read_csv('practice.csv')

duplicate_count =df['EmployeeID'].duplicated().sum()
print(f'duplicate_count: {duplicate_count}')

cleaned_df = df.drop_duplicates(subset='EmployeeID')

cleaned_df
```

# Problem 03. Handle Missing Values  
Columns: Age, Salary   
Tasks:  
Count missing values  
Fill missing Age with mean age  
Drop rows where Salary is missing  


```python
df = pd.read_csv('practice.csv')
nul_count = df[['Age','Salary']].isnull().sum()
print(f'total missing values : {nul_count}')

df['Age'] = df['Age'].fillna(df['Age'].mean())
df = df.dropna(subset=['Salary'])

df
```

# Problem 04. Apply Statistical Functions  
Columns: Marks1, Marks2, Marks3  
 Tasks:  
Calculate mean, max for each subject  
Compute total marks per student  
Find standard deviation of total marks


```python
df = pd.read_csv('practice.csv')

mean_value = df[['Marks1', 'Marks2', 'Marks3']].mean()
max_value = df[['Marks1', 'Marks2', 'Marks3']].max()

print(f'Mean values:\n{mean_value}\n')
print(f'Max values:\n{max_value}\n')

df['Total_Marks'] = df[['Marks1', 'Marks2', 'Marks3']].sum(axis=1)  

std_mark = df['Total_Marks'].std()
print(f'Standard deviation of total marks:\n{std_mark:.2f}')
```

# Problem 05. Histogram, Bar, Scatter, Pie
Create:  
Histogram of Age  
Scatter plot (Age vs Salary)  
Scatter plot ( Price vs Discounted Price)  


```python
import matplotlib.pyplot as plt

df = pd.read_csv('practice.csv')

plt.hist(df['Age'], bins=10, edgecolor='black', color = 'red')
plt.title('Age Distribution')
plt.xlabel('Age')
plt.ylabel('Frequency')
plt.show()
```

```python
plt.scatter(df['Age'], df['Salary'], color='red', alpha= 1, edgecolor='black', s=100, marker='*')
plt.title('Age vs Salary')
plt.xlabel('Age')
plt.ylabel('Salary')
plt.show()
```

```python
df = pd.read_csv('discounted_products.csv')

plt.scatter(df['Price'], df['Discount_price'],alpha = 1, color = 'Green', marker= '+')
plt.title('Price Vs Discounted Price')
plt.xlabel('Price')
plt.ylabel('Discounted price')
plt.show()
```
