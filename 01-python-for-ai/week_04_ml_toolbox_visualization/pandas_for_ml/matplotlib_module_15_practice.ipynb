# 2D line plot from pandas data

```python
df = pd.read_csv('enrollment_data.csv')
df
```

```python
plt.xlabel('Years')
plt.ylabel('Count of enrollment')
plt.title('Student enrollment over 8 years')

plt.plot(df['Year'], df['Programming'])

plt.show()
```

```python
plt.xlabel('Years')
plt.ylabel('Count of enrollment')
plt.title('Student enrollment over 8 years')

plt.plot(df['Year'], df['Digital Marketing'])
plt.plot(df['Year'], df['Programming'])
plt.show()
```

```python
# adding labels using plt.legend()

plt.xlabel('Years')
plt.ylabel('Count of enrollment')
plt.title('Student enrollment over 8 years')

plt.plot(df['Year'], df['Digital Marketing'], label = 'Digital marketing')
plt.plot(df['Year'], df['Programming'], label = 'Programming')

plt.legend()
plt.show()
```

```python
# adding some visualization 

plt.xlabel('Years') 
plt.ylabel('Count of Enrollment')
plt.title('Student Enrollment over 8 years')

plt.plot(df['Year'] ,df['Programming'],label = 'Programming', color ='red',linewidth=3,linestyle='dashed',marker='o',markersize=7)
plt.plot(df['Year'] ,df['Digital Marketing'], label= 'Digital Marketing', color='black',linewidth=3,linestyle='dashed' )

plt.legend() 
plt.show()
```

# Scatter Plot

```python
students = pd.read_csv('student_IQdata.csv')
students
```

```python
# Relation between study Hours and IQ score - postive correlation

plt.xlabel('Study Hours')
plt.ylabel('IQ Score')
plt.title('Relation betweent study hours an IQ score')

plt.scatter(students['Study_Hour'], students['IQ_Score'])
```

```python
# Relation between shoe size and IQ score -> no correlation

plt.xlabel('Shoe size')
plt.ylabel('IQ Score')
plt.title('Relation betweent shoe size and IQ score')

plt.scatter(students['Shoe_Size'], students['IQ_Score'])
```

```python
# Relation between chilling hour and IQ score  -> negative correlation

plt.xlabel('Chilling Time')
plt.ylabel('IQ Score')
plt.title('Relation betweent chilling houre and IQ score')

plt.scatter(students['Chilling_Hours'], students['IQ_Score'], marker = '*', color = 'red')
```

# Histogram

# Bar Chart

```python

data = {
    'Student': [ 'B', 'C', 'D', 'E', 'F', 'G', 'H'],
    'Gender': [ 'Male', 'Female', 'Male', 'Female', 'Male', 'Female', 'Male'],
    'StudyHours': [ 2, 5, 3, 7, 4, 8, 1],
    'Attendance': [ 60, 88, 70, 95, 80, 98, 55],
    'Grade': [ 'Fail', 'Pass', 'Fail', 'Pass', 'Pass', 'Pass', 'Fail']
}

df = pd.DataFrame(data)
```

```python
# gender wise

gender_group = df.groupby('Gender').size()

gender_group.index
gender_group.values
```

```python
plt.bar(gender_group.index, gender_group.values, color = ['orange', 'red'], edgecolor = 'black')

plt.xlabel('Gender')
plt.ylabel('Count')
plt.show()
```

```python
df = pd.read_csv('student_data.csv')
df.head()
```

```python
status = df.groupby('CompletionStatus').size()
plt.bar(status.index, status.values, color=['green', 'orange', 'red'], edgecolor='black')

plt.xlabel('Completion status')
plt.ylabel('Count')
plt.title('Bar chart of course completion')
plt.show()
```

```python
status = df.groupby('Location').size()

plt.bar(status.index, status.values, edgecolor = 'red')
plt.xlabel('Location')
plt.ylabel('Count')
plt.title('Bar chart of the location students')

plt.show()
```

# Pie chart

```python
status = df.groupby('CompletionStatus').size()

plt.pie(status,labels= status.index, autopct='%1.1f%%')
plt.show()
```

```python
status = df.groupby('CompletionStatus').size()

plt.pie(status,labels= status.index, autopct='%1.1f%%', explode= (0.1,0,0),shadow= True)
plt.show()
```
