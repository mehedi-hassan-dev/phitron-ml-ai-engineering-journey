```python
dates = pd.date_range(start="2026-10-01", periods=10, freq="D")

data = {
    "Student": ["A", "B", "C", "D", "E", "F", "G", "H", "I", "J"],
    "Marks": [85, 72, 90, 65, 78, 88, 95, 60, 82, 74],
    "Attendance": [92, 85, 98, 70, 88, 95, 100, 65, 90, 80]
}

date_df = pd.DataFrame(data, index=dates)
date_df.index.name = "Date"

date_df
```

## Value assign kora

```python
copy_df = df.copy() # dataframe er copy kora hoyese
copy_df['gpa'] = 'A'  # puro column ea value assin kore
copy_df.head(4)
```

```python
copy_df["age"] = np.where(df["age"] >= 20, "Adult", "Young")  # condition anujai value assing

copy_df.sample(5)
```

```python
copy_df.loc[0, 'name'] = 'Mehedi'  # akhane row er specific value change kora hoyse --label diye
 
copy_df.head(3)
```

Specific cell ea value change — position diyea

```python
copy_df.iloc[1, copy_df.columns.get_loc("department")] = 'CST'

copy_df.head(3)
```

```python
copy2_df = df.copy()
copy2_df['age'] = copy2_df['age'] + 10   # akhane age column er age er sathe 10 add korbe

copy2_df.head(4)
```

```python
copy2_df["result"] = np.where(copy2_df["gpa"] >= 3, "Pass", "Fail")   # new column add kora

copy2_df.head(4)
```

## Condition use kore filtering

```python
gpa = df[df["gpa"] > 3]   # gpa 3 er ceye beshi segulo show korbe

gpa.head(4)
```

```python
gpa2 = df[(df["gpa"] > 3) & (df["gpa"] > 2.5)]   # multiple condition 'AND' use

gpa2.head(4)
```

```python
age = df[(df["age"] > 33) | (df["age"] > 20)]   # multiple condition 'OR' use kore

age.head(5)
```

```python
age2 = df[~(df["age"] > 19)]   # NOT condition use kor

age2.head(5)
```

```python
name = df[df["name"].isin(["David Palmer", "Rahim"])]  # Specific values thake row dekhabe

name
```

.loc diyea Boolean filtering kora jai

```python
gp = df.loc[df["gpa"] > 3.96]

gp
```
