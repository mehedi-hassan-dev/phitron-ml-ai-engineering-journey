# ndarray & Attributes

```python
a = np.array([[10, 20, 30],
              [40, 50, 60]])

print(a)
print("Dimension:", a.ndim)
print("Shape:", a.shape)
print("Total elements:", a.size)
print("Data type:", a.dtype)
print("Item size:", a.itemsize, "bytes")
```

ndarray Data Type (dtype)

```python
a = np.array([1, 2, 3])
print(a.dtype)

b = np.array([1, 2, 3], dtype=np.float64) # float ea convert
print(b)
print(b.dtype)

c = np.array([1.2, 2.8, 3.5], dtype=np.int32) # float ke integer ea convert
print(c)
print(c.dtype)
```

astype()  >-  use kore data type change kora

```python
numbers = np.array([1.5, 2.7, 3.9])

integer_numbers = numbers.astype(int)
print(integer_numbers)
```

 # Existing Data theke ndarray create

```python
# list theke 1D array create
a = np.array([10, 20, 30, 40])

print(a)


# Nested list theke 2D array create
b = np.array([
    [1, 2, 3],
    [4, 5, 6]
])

print(b)


# Tuple theke array create
c = np.array((100, 200, 300))

print(c)
```

Special arrays

```python
zeros = np.zeros((2, 3))
print(zeros)

ones = np.ones((3, 2), dtype=int)
print(ones)

full_array = np.full((2, 3), 7)
print(full_array)
```

 # Random Values use kore ndarray

```python
np.random.seed(42)

# same value thakbe
random_float = np.random.rand(22)
print(random_float)

x = np.random.randint(100, size=(3, 5))
print(x)
```

Random value generate korer best practice

```python
# Same value bar bar pete ei method use kora hoi 

rng = np.random.default_rng(42) 

random_float = rng.random((2, 3))
print(random_float)


random_int = rng.integers(1, 11, size=(3, 4)) # 1 to 11 range er modho integer array generate kore dibe
print(random_int)
```

# Indexing & Slicing

```python
a = np.array([10, 20, 30, 40, 50])

print(a[0])
print(a[2])
print(a[-1])
```

1D indexing

```python
b = np.array([10, 20, 30, 40, 50])

print(b[1:4])
print(b[:3])
print(b[2:])
print(b[::2])
print(b[::-1])
```

2D indexing

```python
matrix = np.array([
    [10, 20, 30],
    [40, 50, 60],
    [70, 80, 90]
])

print(matrix.ndim)

print(matrix[0, 0])
print(matrix[1, 2])
print(matrix[2, 1])

```

2D slicing

```python
matrix = np.array([
    [10, 20, 30],
    [40, 50, 60],
    [70, 80, 90]
])

print(matrix[0:2, 1:3]) # [row_start:row_end, col_start:col_end]
print(matrix[0, :]) # row slicing
print(matrix[:, 1]) # column slicing
```

# Copy & View

```python
# View 

original = np.array([10, 20, 30, 40, 50])

view_array = original[1:4] 
view_array[0] = 999

print("Original:", original)
print("View:", view_array)
```

```python
#independent copy:

original = np.array([10, 20, 30, 40, 50])

copied_array = original[1:4].copy()
copied_array[0] = 999

print("Original:", original)
print("Copy:", copied_array)
```

 # Advanced Indexing

```python
a = np.array([10, 20, 30, 40, 50, 60])

print(a[2])
print(a[[0, 2, 5]])
```

2D advanced indexing

```python
matrix = np.array([
    [10, 20, 30],
    [40, 50, 60],
    [70, 80, 90]
])

print(matrix[[0, 2], [1, 2]]) 
```

Boolean indexing

```python
marks = np.array([45, 72, 88, 33, 60, 91])

passed_marks = marks[marks >= 60]
print(passed_marks)


even_numbers = np.array([1, 2, 3, 4, 5, 6, 7, 8])
print(even_numbers[even_numbers % 2 == 0])
```

# ndarray Iteration

```python
a = np.array([10, 20, 30])

for value in a:
    print(value)
```

2D array iteration

```python
matrix = np.array([
    [1, 2, 3],
    [4, 5, 6]
])

for row in matrix:
    print(row)

# Nested for loop use kore item print

for x in matrix:
    for z in x:
        print(z)
```

Index soho iterate korte

```python
matrix = np.array([
    [1, 2, 3],
    [4, 5, 6]
])

for index, value in np.ndenumerate(matrix):
    print(index, value)
```
