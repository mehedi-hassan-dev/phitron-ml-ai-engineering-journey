## ndarray manipulation

```python
import numpy as np
```

```python
a = np.arange(1, 13)

print(a)
print(a.shape)

b = a.reshape(3, 4)

print(b)
print(b.shape)

```

flatten() and ravel() 

```python
matrix = np.array([
    [1, 2, 3],
    [4, 5, 6]
])

flat = matrix.flatten()  # generally copy dey

print(matrix.ndim)
print(flat)
print(flat.ndim)


ravel_array = matrix.ravel()  # generally view dey

print(ravel_array)
```

transpose() method

```python
matrix = np.array([
    [1, 2, 3],
    [4, 5, 6]
])

print(matrix.T)
```

array concatenate()

```python
a = np.array([1, 2, 3])
b = np.array([4, 5, 6])

result = np.concatenate((a, b))

print(result)
```

```python
# 2D array concatenate

x = np.array([[1, 2], [3, 4]])
y = np.array([[5, 6], [7, 8]])

print(np.concatenate((x, y), axis=0)) # row waise add korbe
print(np.concatenate((x, y), axis=1))  # column waise add korbe

print(np.vstack((x, y))) #  niche jora dey
print(np.hstack((x, y))) #  pase jora dey
```

split() mathod

```python
a = np.arange(1, 13)

parts = np.split(a, 3)

print(parts)


matrix = np.arange(1, 13).reshape(3, 4)

print(np.hsplit(matrix, 2))
print(np.vsplit(matrix, 3))
```

```python
a = np.array([1, 2, 3, 4])

print(np.resize(a, (3, 3)))
```

## Arithmetic Operators ও Mathematical Functions

```python
a = np.array([10, 20, 30])
b = np.array([1, 2, 3])

print(a + b)
print(a - b)
print(a * b)
print(a / b)
print(a // b)
print(a % b)
print(a ** b)

print(a + 5)
print(a * 2)
```

Important mathematical function

```python

numbers = np.array([1, 4, 9, 16, 25])

print(np.sqrt(numbers))
print(np.square(numbers))
print(np.power(numbers, 2))
print(np.abs(np.array([-5, -2, 0, 3])))
```

Rounding functions

```python
values = np.array([1.2, 1.5, 1.8, 2.7])

print(np.round(values))
print(np.floor(values))
print(np.ceil(values))
```

## NumPy Logical Functions

```python
marks = np.array([45, 72, 88, 33, 60])

print(marks > 50)
print(marks == 60)
print(marks != 45)
```

logical_and(), logical_or(), logical_not() methods

```python
marks = np.array([45, 72, 88, 33, 60])

condition = np.logical_and(marks >= 50, marks <= 80)

print(condition)
print(marks[condition])

condition = np.logical_or(marks < 40, marks > 80)
print(marks[condition])

condition = marks >= 60
print(np.logical_not(condition))
```

any()  &  all()

```python
marks = np.array([45, 72, 88, 33, 60])

print(np.any(marks >= 80))
print(np.all(marks >= 40))
```

## ndarray Sorting

```python
a = np.array([50, 10, 40, 20, 30])

sorted_a = np.sort(a)

print(sorted_a) # orginal array te change ne kore sort kore
print(a)

a.sort() # orginal array te change kore sort kore
print(a)
```

2D sorting

```python
matrix = np.array([
    [30, 70, 20],
    [60, 10, 50]
])

print(np.sort(matrix, axis=0))  # column wise sort kore

print(np.sort(matrix, axis=1))  # row wise sort kore
```

Sorted order er index jante argsort() use kora hoi

```python
marks = np.array([72, 45, 88, 60])

indexes = np.argsort(marks)

print(indexes)
print(marks[indexes])
```

 ## ndarray Searching

```python
marks = np.array([45, 72, 88, 33, 60])

indexes = np.where(marks >= 60)

print(indexes)
print(marks[indexes])
```

Condition use kore value replace

```python
marks = np.array([45, 72, 88, 33, 60])

result = np.where(marks >= 50, "Pass", "Fail")

print(result)
```

searchsorted()

```python
a = np.array([10, 20, 30, 40, 50])

position = np.searchsorted(a, 35)

print(position)
```

## ndarray Counting

```python
marks = np.array([45, 72, 88, 33, 60, 72])

print(np.count_nonzero(marks >= 60))
```

unique() & frequency count

```python
numbers = np.array([1, 2, 2, 3, 3, 3, 4, 2, 4, 1, 2, ])

unique_values, counts = np.unique(numbers, return_counts=True)

print(unique_values)
print(counts)
```

bincount()

```python
numbers = np.array([0, 0, 1, 1, 2, 2, 2, 4, 6, 7, 7])

print(np.bincount(numbers))
```

## Statistical Functions

```python
marks = np.array([60, 70, 80, 90, 100])

print("Sum:", np.sum(marks))
print("Mean:", np.mean(marks))
print("Median:", np.median(marks))
print("Minimum:", np.min(marks))
print("Maximum:", np.max(marks))
print("Range:", np.ptp(marks))
print("Variance:", np.var(marks))
print("Standard Deviation:", np.std(marks))
```

Axis anujai statistics

```python
matrix = np.array([
    [10, 20, 30],
    [40, 50, 60]
])

print("Column sum:", np.sum(matrix, axis=0))
print("Row sum:", np.sum(matrix, axis=1))
print("Column mean:", np.mean(matrix, axis=0))
print("Row mean:", np.mean(matrix, axis=1))
```

## Linear Algebra with NumPy

Matrix addition & subtraction

```python
A = np.array([
    [1, 2],
    [3, 4]
])

B = np.array([
    [5, 6],
    [7, 8]
])

print(A + B)
print(A - B)
print(A * B) # element wise multiplication kore
print(np.dot(A,B))  # matrix multiplication kore
print(A @ B) # same matrix multiplication kore
```

Determinant

```python
A = np.array([
    [1, 2],
    [3, 4]
])

determinant = np.linalg.det(A)

print(determinant)
```
